```python
import re

import requests
from bs4 import BeautifulSoup, ResultSet
```

```python
USER_AGENT: str = "Mozilla/5.0 (Linux; Android 15; Pixel 9) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Mobile Safari/537.36"

CUSTOM_HEADERS: dict[str, str] = {
    "User-Agent": USER_AGENT,
    "accept-language": "en-US,en;q=0.9",
    "accept-encoding": "gzip, deflate, br",
}

BASE_SEARCH_URL: str = "https://www.amazon.in/s"

search_item: str = "laptop"
```

```python
search_query_params: dict[str, str | int] = {
    "k": search_item,
    "page": 1,
}


site_response: requests.Response = requests.get(
    BASE_SEARCH_URL, params=search_query_params, headers=CUSTOM_HEADERS
)

site_response_status_code: int = site_response.status_code

if site_response_status_code != 200:
    print(
        "Something went wrong, failed to get 200 response from the", site_response.url
    )
else:
    print("Ok:", site_response.url)
```

```python
max_search_page_count: int = 50

for page in range(1, max_search_page_count + 1):
    search_query_params["page"] = page

    response = requests.get(
        BASE_SEARCH_URL, params=search_query_params, headers=CUSTOM_HEADERS
    )

    if site_response_status_code != 200:
        print("Faild to get 200 response from the", response.url, "Page:", page)

        # Reset search_query_params back to default
        search_query_params["page"] = 1
        break
    else:
        print("Ok:", response.url)

    # Reset search_query_params back to default on reaching max_search_page_count
    if page == max_search_page_count:
        search_query_params["page"] = 1

    soup: BeautifulSoup = BeautifulSoup(response.content, "html.parser")

    product_containers: ResultSet = soup.find_all("div", class_="s-result-item")

    for container in product_containers:
        # Step 1: Extract title from h2 tag
        h2_tag = container.find("h2")
        
        if not h2_tag:
            continue

        title = h2_tag.text.strip()



        # Extract title
        brand_re_pattern = re.search(r"^([A-Za-z]+)", title)
        brand = brand_re_pattern.group(1) if brand_re_pattern else "Unknown"

        # Extract the ram from the title using regex GB,ram DDR LPDDR
        match = re.search(
            r"(\d+GB|RAM\s*(\d+GB)?|DDR\d?\s*(\d+GB)?|LPDDR\d?\s*(\d+GB)?)",
            title,
            re.IGNORECASE,
        )
        ram = match.group(1) if match else "N/A"

        match = re.search(
            r"(\d+)\s*(GB|TB)\s*(?:SSD|Storage|HDD)", title, re.IGNORECASE
        )
        storage = f"{match.group(1)}{match.group(2)}" if match else "N/A"

        match = re.search(
            r"\b(Black|White|Silver|Gray|Grey|Red|Blue|Green|Yellow|Pink|Purple|Gold|Bronze|Rose Gold)\b",
            title,
            re.IGNORECASE,
        )
        color = match.group(1) if match else "N/A"

```
