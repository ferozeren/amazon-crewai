```python
from bs4 import BeautifulSoup, Tag
import requests
import pandas as pd


BASE_URL: str = "https://quotes.toscrape.com"

USER_AGENT: str = "Mozilla/5.0 (Linux; Android 15; Pixel 9) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Mobile Safari/537.36"

CUSTOM_HEADERS: dict[str, str] = {
    "User-Agent": USER_AGENT,
    "accept-language": "en-US,en;q=0.9",
    "accept-encoding": "gzip, deflate, br",
}


max_page_count: int = 10

list_text : list[str] = []
list_author : list[str] = []
list_tags_list : list[list[str]] = []


for page in range(1, max_page_count + 1):
    page_no: str = f"/page/{page}"

    site_response: requests.Response = requests.get(
        BASE_URL + page_no, headers=CUSTOM_HEADERS
    )

    if site_response.status_code == 200:
        print("Ok:", site_response.url)
    else:
        print("STATUS CODE",site_response.status_code ,"Failed to get response form the", site_response.url)
        break

    soup = BeautifulSoup(site_response.content, "html.parser")

    quotes = soup.find_all("div", class_='quote')

    for quote in quotes:
        quote_text: Tag | None = quote.find("span", class_="text")
        quote_author: Tag | None = quote.find("small", class_="author")
       

        if not (quote_text and quote_author):
            print("Continue")
            continue
        list_text.append(quote_text.text)
        list_author.append(quote_author.text)

        tag_container = quote.find_all("div", class_="tags")

        tags_list: list[str] = []

        for quote_tags in tag_container:
            tags  = quote_tags.find_all("a", class_="tag")

            for tag in tags:
                tags_list.append(tag.text)
        
        list_tags_list.append(tags_list)
        
```

```python
if len(list_tags_list) == len(list_author) == len(list_text):
    print("Scrapped successfully")
    print(list_text, list_author, list_tags_list, sep="\n")
else:
    print("Scrapped had irregularities")
    print(len(list_text), len(list_author), len(list_tags_list))

print(list_author)


```
